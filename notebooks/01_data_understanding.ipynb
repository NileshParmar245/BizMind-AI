```python
# Create a copy of the raw dataset
clean_df = df.copy()

# 1. Remove exact duplicate rows
clean_df = clean_df.drop_duplicates()

# 2. Remove rows with missing Description
clean_df = clean_df.dropna(subset=["Description"])

# 3. Remove invalid UnitPrice values
# Keep only transactions with positive prices
clean_df = clean_df[clean_df["UnitPrice"] > 0]

# 4. Create a cancellation flag
clean_df["IsCancelled"] = (
    clean_df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

# 5. Create TotalAmount
clean_df["TotalAmount"] = (
    clean_df["Quantity"] * clean_df["UnitPrice"]
)

# Display cleaned dataset information
print("Cleaned dataset created successfully!")
print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))

print("\nCleaned dataset columns:")
print(clean_df.columns.tolist())

display(clean_df.head(10))
```

```python
# Feature Engineering

# Make sure InvoiceDate is datetime
clean_df["InvoiceDate"] = pd.to_datetime(clean_df["InvoiceDate"])

# Extract date/time features
clean_df["Year"] = clean_df["InvoiceDate"].dt.year
clean_df["Month"] = clean_df["InvoiceDate"].dt.month
clean_df["MonthName"] = clean_df["InvoiceDate"].dt.month_name()
clean_df["Day"] = clean_df["InvoiceDate"].dt.day
clean_df["DayName"] = clean_df["InvoiceDate"].dt.day_name()
clean_df["Hour"] = clean_df["InvoiceDate"].dt.hour

# Customer type information
clean_df["CustomerID"] = clean_df["CustomerID"].fillna(0)

# Create a customer identifier category
clean_df["CustomerType"] = np.where(
    clean_df["CustomerID"] == 0,
    "Guest",
    "Registered"
)

# Display the new columns
print("Feature engineering completed!")

print("\nNew columns:")
print([
    "Year",
    "Month",
    "MonthName",
    "Day",
    "DayName",
    "Hour",
    "CustomerType"
])

display(clean_df.head(10))
```

```python
# RFM Customer Segmentation

# Reference date = day after the latest transaction
reference_date = clean_df["InvoiceDate"].max() + pd.Timedelta(days=1)

rfm = (
    clean_df.dropna(subset=["CustomerID"])
    .groupby("CustomerID")
    .agg(
        Recency=("InvoiceDate", lambda x: (reference_date - x.max()).days),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("TotalRevenue", "sum")
    )
    .reset_index()
)

print("RFM Customer Analysis:")
display(rfm.head(10))

print("\nRFM Summary:")
display(rfm[["Recency", "Frequency", "Monetary"]].describe())
```

```python
# Clean RFM data before scoring

rfm_clean = rfm[
    (rfm["CustomerID"] > 0) &
    (rfm["Monetary"] > 0) &
    (rfm["Frequency"] > 0)
].copy()

print("RFM customers after cleaning:", len(rfm_clean))

print("\nRFM Data:")
display(rfm_clean.head(10))

print("\nRFM Summary:")
display(rfm_clean[["Recency", "Frequency", "Monetary"]].describe())
```

```python
# ============================================
# RFM SCORING
# ============================================

# Recency:
# Lower recency = better customer
rfm_clean["R_Score"] = pd.qcut(
    rfm_clean["Recency"],
    q=5,
    labels=[5, 4, 3, 2, 1],
    duplicates="drop"
).astype(int)

# Frequency:
# Higher frequency = better customer
rfm_clean["F_Score"] = pd.qcut(
    rfm_clean["Frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary:
# Higher spending = better customer
rfm_clean["M_Score"] = pd.qcut(
    rfm_clean["Monetary"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Combined RFM Score
rfm_clean["RFM_Score"] = (
    rfm_clean["R_Score"].astype(str) +
    rfm_clean["F_Score"].astype(str) +
    rfm_clean["M_Score"].astype(str)
)

print("RFM scoring completed successfully!")

display(
    rfm_clean[
        [
            "CustomerID",
            "Recency",
            "Frequency",
            "Monetary",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score"
        ]
    ].head(20)
)
```

```python
# ============================================
# CUSTOMER SEGMENTATION
# ============================================

def segment_customer(row):
    r = row["R_Score"]
    f = row["F_Score"]
    m = row["M_Score"]

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"

    elif r >= 4 and f >= 3:
        return "Loyal Customers"

    elif r >= 4 and f <= 2:
        return "New Customers"

    elif r == 3 and f >= 3:
        return "Potential Loyalists"

    elif r <= 2 and f >= 4:
        return "At Risk"

    elif r <= 2 and f <= 2 and m >= 3:
        return "Cannot Lose Them"

    elif r <= 2 and f <= 2:
        return "Lost Customers"

    else:
        return "Needs Attention"


# Apply segmentation
rfm_clean["CustomerSegment"] = rfm_clean.apply(
    segment_customer,
    axis=1
)

print("Customer segmentation completed!")

# Display sample
display(
    rfm_clean[
        [
            "CustomerID",
            "Recency",
            "Frequency",
            "Monetary",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score",
            "CustomerSegment"
        ]
    ].head(20)
)
```

```python
# ============================================
# RFM SEGMENT SUMMARY
# ============================================

segment_summary = (
    rfm_clean
    .groupby("CustomerSegment")
    .agg(
        Customers=("CustomerID", "count"),
        AvgRecency=("Recency", "mean"),
        AvgFrequency=("Frequency", "mean"),
        AvgMonetary=("Monetary", "mean"),
        TotalRevenue=("Monetary", "sum")
    )
    .sort_values("TotalRevenue", ascending=False)
    .reset_index()
)

# Round values for cleaner output
segment_summary["AvgRecency"] = segment_summary["AvgRecency"].round(1)
segment_summary["AvgFrequency"] = segment_summary["AvgFrequency"].round(1)
segment_summary["AvgMonetary"] = segment_summary["AvgMonetary"].round(2)
segment_summary["TotalRevenue"] = segment_summary["TotalRevenue"].round(2)

print("RFM CUSTOMER SEGMENT SUMMARY")
display(segment_summary)
```

```python
# ============================================
# CUSTOMER COUNT BY RFM SEGMENT
# ============================================

plt.figure(figsize=(12, 6))

segment_counts = (
    rfm_clean["CustomerSegment"]
    .value_counts()
    .sort_values(ascending=False)
)

plt.bar(
    segment_counts.index,
    segment_counts.values
)

plt.title("Customers by RFM Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=35, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================
# REVENUE BY RFM CUSTOMER SEGMENT
# ============================================

segment_revenue = (
    rfm_clean
    .groupby("CustomerSegment")["Monetary"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by RFM Segment:")
display(segment_revenue.reset_index(name="TotalRevenue"))

# Plot
plt.figure(figsize=(12, 6))

plt.bar(
    segment_revenue.index,
    segment_revenue.values
)

plt.title("Revenue by RFM Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Revenue (£)")
plt.xticks(rotation=35, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================
# REVENUE CONTRIBUTION BY RFM SEGMENT
# ============================================

segment_revenue_df = (
    rfm_clean
    .groupby("CustomerSegment")["Monetary"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

# Calculate percentage contribution
total_segment_revenue = segment_revenue_df["Monetary"].sum()

segment_revenue_df["RevenuePercentage"] = (
    segment_revenue_df["Monetary"] /
    total_segment_revenue * 100
).round(2)

# Rename column
segment_revenue_df = segment_revenue_df.rename(
    columns={"Monetary": "TotalRevenue"}
)

print("Revenue Contribution by RFM Segment:")
display(segment_revenue_df)
```

```python
# ============================================
# RFM BUSINESS RECOMMENDATIONS
# ============================================

segment_strategy = pd.DataFrame({
    "CustomerSegment": [
        "Champions",
        "Potential Loyalists",
        "At Risk",
        "Needs Attention",
        "Loyal Customers",
        "Cannot Lose Them",
        "Lost Customers",
        "New Customers"
    ],
    
    "Strategy": [
        "Reward loyalty, exclusive offers and early access",
        "Upselling, cross-selling and loyalty programs",
        "Win-back campaigns and personalized discounts",
        "Targeted engagement and product recommendations",
        "Loyalty rewards and membership benefits",
        "High-priority retention and personalized offers",
        "Reactivation campaigns and special discounts",
        "Welcome offers and encourage second purchase"
    ]
})

print("RFM BUSINESS STRATEGY:")
display(segment_strategy)
```

```python
# ============================================
# RFM CORRELATION ANALYSIS
# ============================================

# Select RFM metrics
rfm_corr = rfm_clean[
    ["Recency", "Frequency", "Monetary"]
].corr()

print("RFM Correlation Matrix:")
display(rfm_corr.round(2))

# Correlation heatmap
plt.figure(figsize=(8, 6))

plt.imshow(rfm_corr, cmap="Blues", vmin=-1, vmax=1)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(rfm_corr.columns)),
    rfm_corr.columns
)

plt.yticks(
    range(len(rfm_corr.columns)),
    rfm_corr.columns
)

plt.title("RFM Feature Correlation Heatmap")

# Add correlation values
for i in range(len(rfm_corr)):
    for j in range(len(rfm_corr)):
        plt.text(
            j,
            i,
            f"{rfm_corr.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()
```

```python
# ============================================
# FINAL RFM SEGMENT PERFORMANCE
# ============================================

rfm_performance = (
    rfm_clean
    .groupby("CustomerSegment")
    .agg(
        Customers=("CustomerID", "count"),
        AvgRecency=("Recency", "mean"),
        AvgFrequency=("Frequency", "mean"),
        AvgMonetary=("Monetary", "mean"),
        TotalRevenue=("Monetary", "sum")
    )
    .reset_index()
)

# Customer percentage
rfm_performance["CustomerPercentage"] = (
    rfm_performance["Customers"] /
    len(rfm_clean) * 100
).round(2)

# Revenue percentage
rfm_performance["RevenuePercentage"] = (
    rfm_performance["TotalRevenue"] /
    rfm_performance["TotalRevenue"].sum() * 100
).round(2)

# Round numerical columns
rfm_performance["AvgRecency"] = rfm_performance["AvgRecency"].round(1)
rfm_performance["AvgFrequency"] = rfm_performance["AvgFrequency"].round(1)
rfm_performance["AvgMonetary"] = rfm_performance["AvgMonetary"].round(2)
rfm_performance["TotalRevenue"] = rfm_performance["TotalRevenue"].round(2)

# Sort by revenue
rfm_performance = rfm_performance.sort_values(
    "TotalRevenue",
    ascending=False
).reset_index(drop=True)

print("FINAL RFM SEGMENT PERFORMANCE")
display(rfm_performance)
```
